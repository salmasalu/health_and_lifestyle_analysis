Introduction:-

This project is based on a health and lifestyle dataset consisting of 100,000 records and 16 attributes. The dataset brings together information about individuals’ demographics, health indicators, and daily lifestyle habits. Key variables include age, gender, body mass index (BMI), daily steps, sleep hours, water intake, calorie consumption, smoking and alcohol habits, along with other wellness-related factors.

The richness of the dataset lies in its combination of numerical and categorical variables. Numerical attributes like age, BMI, steps, and sleep hours allow statistical analysis of trends and distributions. Meanwhile, categorical attributes such as gender, smoker, and alcohol status enable comparisons between groups. Together, these features provide a comprehensive picture of wellness behaviors across a large and diverse sample of individuals.

The goal of this project is to analyze patterns, trends, and relationships between lifestyle choices and health outcomes, which can provide valuable insights into wellness and healthy living and it valuable for understanding how lifestyle choices impact health outcomes.

Outlier Detection

```python
#Handle outliers
numeric_cols = df.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    df = df[(df[col] >= Q1 - 1.5 * IQR) & (df[col] <= Q3 + 1.5 * IQR)]

# Boxplot for numeric columns
df.boxplot(figsize=(15, 8))
plt.xticks(rotation=45)
plt.title("Boxplot of Numeric Columns After Outlier Removal")
plt.show()

```

Feature Engineering


```python
#1.Body Efficiency
df['Efficiency_steps_per_calorie'] = df['daily_steps'] / df['calories_consumed']

#2.Hydration Efficiency
df['Hydration_per_BMI'] = df['water_intake_l'] / df['bmi']

#3.Sleep Quality Index
df['Sleep_Quality_Index'] = df['sleep_hours'] / df['resting_hr']

#4.Blood Pressure Ratio
df['BP_Ratio'] = df['systolic_bp'] / df['diastolic_bp']

# 5.Health Stress Index 
df['Health_Stress_Index'] = (df['cholesterol'] * df['resting_hr']) / 1000

#6.Lifestyle Score
df['Lifestyle_Score'] = (
    (df['daily_steps'] / df['daily_steps'].max()) * 0.3 +
    (df['sleep_hours'] / df['sleep_hours'].max()) * 0.2 +
    (df['water_intake_l'] / df['water_intake_l'].max()) * 0.2 -
    (df['calories_consumed'] / df['calories_consumed'].max()) * 0.2 -
    (df['smoker'] * 0.05 + df['alcohol'] * 0.05))
#7. Age Grouping — young, middle, senior
df['Age_Group'] = pd.cut(df['age'], bins=[0, 30, 50, 70, 100], labels=['Young', 'Middle-aged', 'Senior', 'Elderly'])

#Preview newly created features
print(df[['Efficiency_steps_per_calorie', 'Hydration_per_BMI', 'Sleep_Quality_Index',
          'BP_Ratio', 'Health_Stress_Index', 'Lifestyle_Score', 'Age_Group']].head())

```
